```python
def loss_func(w):
    return (w*w*)
```

```python
loss_func(2)
```

```python
w = 2

loss_func(w)
```

```python
def derivative(loss_func, w):

  delta_w = 0.000001

  return (loss_func(w+delta_w)-loss_func(w))/delta_w
```

```python
derivative(loss_func, 2)
```

```python
def gradient_descent(loss_func, lr, w, epochs):

  for i in range(epochs):
    w = w - lr*derivative(loss_func, w)
    print("weight is ", w)
    print("loss is ", loss_func(w))
  return w
```

```python
gradient_descent(loss_func,0.1, -3, 20)
```
