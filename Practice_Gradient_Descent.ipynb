```python
def loss_fn(w):
  return w*w

def derivative(loss_fn,w):
  dw=0.001
  return (loss_fn(w+dw)-loss_fn(w))/dw

#derivative(loss_fn,3)

def gradient_d_epoch_controlled(loss_fn,w,lr,epochs):
  for i in range(epochs):
    w=w-lr*derivative(loss_fn,w)
    print("New weight : ",w)
    print("New Loss : ",loss_fn(w))

def gradient_d_intelligent(loss_fn,w,lr):
  acceptable_loss=0.02
  while loss_fn(w)>acceptable_loss:
    w=w-lr*derivative(loss_fn,w)
    print("New weight : ",w)
    print("New Loss : ",loss_fn(w))

gradient_d_intelligent(loss_fn,3,0.1)     ## No Epoch, decides itself when to exit

#gradient_d(loss_fn,3,0.1,25)             ## With Epoch
```
